```python
import requests
from bs4 import BeautifulSoup
import csv
import json
import time
```

```python
# Base URL for Trustpilot reviews page
base_url = 'https://www.trustpilot.com/review/smartbuyglasses.ca'

# Headers to mimic a real browser so the server does not block the request
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/124.0.0.0 Safari/537.36'
}

# List to store all scraped reviews
reviews = []

# Loop through pages 1 to 5 (as required by assignment)
for page_num in range(1, 6):

    # Send a GET request to the current page
    response = requests.get(base_url + '?page=' + str(page_num), headers=headers)

    # Parse the HTML content of the page
    soup = BeautifulSoup(response.text, 'html.parser')

    # Trustpilot stores all review data inside a <script id="__NEXT_DATA__"> tag as JSON
    script_tag = soup.find('script', {'id': '__NEXT_DATA__'})

    # Parse the JSON content from the script tag
    json_data = json.loads(script_tag.string)

    # Navigate into the JSON to get the list of reviews
    page_reviews = json_data['props']['pageProps']['reviews']

    # Extract reviewer name, review date, and review content from each review
    for review in page_reviews:

        # Extract reviewer name
        reviewer_name = review['consumer']['displayName']

        # Extract review date (keep only YYYY-MM-DD from the full timestamp)
        review_date = review['dates']['publishedDate'][:10]

        # Extract the written review content
        review_content = review.get('text', '[No written review]').strip()

        # Print extracted info
        print('Reviewer:', reviewer_name)
        print('Date:', review_date)
        print('Review:', review_content)
        print()

        # Append to reviews list
        reviews.append({
            'reviewer_name': reviewer_name,
            'review_date': review_date,
            'review_content': review_content
        })

    print('--- Page', page_num, 'done:', len(page_reviews), 'reviews scraped ---')
    print()

    # Wait 2 seconds between requests to be polite to the server
    time.sleep(2)

# Save all reviews to a CSV file
with open('trustpilot_reviews.csv', 'w', newline='', encoding='utf-8-sig') as csvfile:
    fieldnames = ['reviewer_name', 'review_date', 'review_content']
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)

    # Write the header row
    writer.writeheader()

    # Write all review rows
    writer.writerows(reviews)

print('Done!', len(reviews), 'reviews saved to trustpilot_reviews.csv')
```
